# Interactive Scatter Analysis

This notebook creates interactive 3D and 2D scatter plots.

```
import plotly.express as px
import numpy as np
import pandas as pd

np.random.seed(42)
df2 = pd.DataFrame({
    'x': np.random.randn(200),
    'y': np.random.randn(200),
    'z': np.random.randn(200),
    'category': np.random.choice(['A','B','C'], size=200)
})

fig = px.scatter_3d(
    df2, x='x', y='y', z='z',
    color='category',
    title='Interactive 3D Scatter by Category'
)
fig.show()
```

```
fig2 = px.scatter(
    df2, x='x', y='y', marginal_x='histogram', marginal_y='box',
    title='Scatter with Marginal Distributions'
)
fig2.show()
```
