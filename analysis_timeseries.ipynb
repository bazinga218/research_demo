# Time Series Analysis

This notebook demonstrates time-series decomposition.

```
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

df = pd.read_csv('data/sample.csv', parse_dates=['date'], index_col='date')

fig, ax = plt.subplots()
df['value'].plot(ax=ax, title='Monthly Values')
ax.set_ylabel('Value')
plt.show()

result = seasonal_decompose(df['value'], model='additive', period=3)
fig = result.plot()
fig.suptitle('Additive Decomposition', y=1.02)
plt.tight_layout()
plt.show()
```

```
df['rolling_mean'] = df['value'].rolling(window=2).mean()
fig, ax = plt.subplots()
df[['value', 'rolling_mean']].plot(ax=ax, title='Value vs. 2-Month Rolling Mean')
ax.set_ylabel('Value')
plt.show()
```
